### 3. Import a complete module

`import math` makes the module available. We refer to its contents with `math.`.


````python
import math

radius = 5
area = math.pi * radius ** 2

print("Pi:", math.pi)
print(f"Circle area: {area:.2f}")
````

### 4. Import with an alias

An alias is a shorter conventional name. `statistics as stats` does not install or change the module.


````python
import statistics as stats

response_times = [12, 15, 13, 18, 14]
print("Mean:", stats.mean(response_times))
print("Median:", stats.median(response_times))
````

### 9. Debug a `TypeError`

A form value may look numerical but still be text. Inspect its type, convert it, then calculate.


````python
price_from_form = "24.95"
quantity = 3

print("Before conversion:", type(price_from_form).__name__)
numeric_price = float(price_from_form)
print("After conversion:", type(numeric_price).__name__)
print("Total:", numeric_price * quantity)
````

### 11. Handle an optional package check

Check availability before importing a package that may not yet be installed.


````python
from importlib.util import find_spec

package_name = "sqlalchemy"
package_available = find_spec(package_name) is not None

print(f"{package_name} available: {package_available}")
if not package_available:
    print("Install it in the active environment, then restart the kernel")
````

### 12. Trace a calculation in stages

Intermediate variables make code easier to inspect than one unexplained expression.


````python
unit_price = 18.00
quantity = 5
discount_rate = 0.10
tax_rate = 0.20

subtotal = unit_price * quantity
discount = subtotal * discount_rate
after_discount = subtotal - discount
tax = after_discount * tax_rate
final_total = after_discount + tax

print("subtotal:", subtotal)
print("discount:", discount)
print("after_discount:", after_discount)
print("tax:", tax)
print("final_total:", final_total)
````

## 13. A professional notebook structure

1. Title and decision/question
2. Data source and definitions
3. Imports and configuration
4. Data checks
5. Analysis
6. Results with units
7. Limitations and conclusion

Markdown should explain **why** a step is performed. Comments should clarify non-obvious code, not repeat every symbol.


### 14. Guided mini-analysis

This example imports a standard module, calculates summaries, and separates a result from its interpretation.


````python
import statistics as stats

weekly_orders = [82, 91, 88, 104, 110]
target = 95

mean_orders = stats.mean(weekly_orders)
median_orders = stats.median(weekly_orders)
target_met = mean_orders >= target

print("Observations:", len(weekly_orders))
print(f"Mean weekly orders: {mean_orders:.1f}")
print(f"Median weekly orders: {median_orders:.1f}")
print("Mean target met:", target_met)
print("Limitation: five observations do not establish a long-term trend")
````
