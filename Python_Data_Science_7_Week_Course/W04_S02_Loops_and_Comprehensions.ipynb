### 1. Repeat with a `for` loop

A `for` loop takes each item from an iterable and runs the indented block.


```python
regions = ["North", "South", "East", "West"]

for region in regions:
    print("Processing:", region)
```

### 2. Accumulate a total

Create an accumulator before the loop and update it once per item.


```python
order_values = [120.0, 85.5, 210.0, 64.5]
total = 0.0

for value in order_values:
    total = total + value

print("Total:", total)
```

### 3. Apply a condition inside a loop

Each value can follow a different branch.


```python
stock_levels = [12, 3, 0, 8, 2]

for stock in stock_levels:
    if stock == 0:
        print(stock, "-> out of stock")
    elif stock < 5:
        print(stock, "-> reorder")
    else:
        print(stock, "-> sufficient")
```

### 4. Use `range`

`range(start, stop, step)` generates integers up to but excluding the stop value.


```python
print("0 to 4:")
for number in range(5):
    print(number)

print("2 to 10 by 2:")
for number in range(2, 11, 2):
    print(number)
```

### 5. Use `enumerate`

`enumerate` supplies both a position and item. `start=1` is useful for human-facing numbering.


```python
products = ["Keyboard", "Monitor", "Mouse"]

for number, product in enumerate(products, start=1):
    print(f"{number}. {product}")
```

### 6. Use `zip`

`zip` pairs items by position. It stops when the shortest input ends.


```python
products = ["Keyboard", "Monitor", "Mouse"]
prices = [45.0, 210.0, 18.0]

for product, price in zip(products, prices):
    print(f"{product}: £{price:.2f}")
```

### 7. Iterate through a dictionary

`.items()` provides a key and value on each iteration.


```python
inventory = {"A10": 8, "B20": 0, "C30": 14}

for sku, quantity in inventory.items():
    print(f"{sku}: {quantity} units")
```

### 8. Build a result list

Begin with an empty list and append one derived value per iteration.


```python
prices = [10.0, 20.0, 30.0]
prices_with_tax = []

for price in prices:
    taxed_price = price * 1.20
    prices_with_tax.append(taxed_price)

print(prices_with_tax)
```

### 9. `continue` skips one iteration

Use `continue` when the current record should be skipped but processing should continue.


```python
values = [120, None, 85, None, 210]
total = 0

for value in values:
    if value is None:
        continue
    total = total + value

print("Known-value total:", total)
```

### 10. `break` stops a loop

Use `break` when further processing is inappropriate after a condition is met.


```python
transactions = [120, 85, -999, 210, 64]
reviewed_total = 0

for transaction in transactions:
    if transaction < -100:
        print("Large negative value found; manual review required")
        break
    reviewed_total = reviewed_total + transaction

print("Total before stop:", reviewed_total)
```

### 11. A basic `while` loop

A `while` loop repeats while its condition is true. The body must change something that eventually makes the condition false.


```python
count = 1

while count <= 5:
    print("Count:", count)
    count = count + 1

print("Loop finished")
```

### 12. A controlled savings loop

A maximum-iteration condition protects against an accidental endless loop.


```python
balance = 1000.0
target = 1500.0
monthly_contribution = 100.0
month = 0
maximum_months = 24

while balance < target and month < maximum_months:
    balance = balance + monthly_contribution
    month = month + 1

print("Months:", month)
print("Final balance:", balance)
print("Target reached:", balance >= target)
```

### 13. List comprehensions

A comprehension is concise for one simple transformation or filter. Use a normal loop for complex logic.


```python
numbers = [1, 2, 3, 4, 5, 6]
squares = [number ** 2 for number in numbers]
even_numbers = [number for number in numbers if number % 2 == 0]
even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print("Squares:", squares)
print("Even numbers:", even_numbers)
print("Even squares:", even_squares)
```

### 14. Dictionary comprehensions

A dictionary comprehension creates one key–value pair per iteration.


```python
products = ["Notebook", "Pen", "Folder"]
prices = [8.0, 1.5, 4.0]

price_lookup = {product: price for product, price in zip(products, prices)}
print(price_lookup)
```

### 15. Put a loop inside a function

A function makes repeated processing reusable and testable.


```python
def count_values_above(values, threshold):
    count = 0
    for value in values:
        if value > threshold:
            count = count + 1
    return count

sales = [82, 105, 99, 140, 76, 121]
print("Above 100:", count_values_above(sales, 100))
print("Above 80:", count_values_above(sales, 80))
```

### 16. Guided customer segmentation

Read each record, calculate a segment, copy the record, and store the result.


```python
customers = [
    {"id": "C1", "annual_spend": 1250},
    {"id": "C2", "annual_spend": 480},
    {"id": "C3", "annual_spend": 750},
    {"id": "C4", "annual_spend": 2100},
]
segmented_customers = []

for customer in customers:
    spend = customer["annual_spend"]
    if spend >= 1000:
        segment = "High"
    elif spend >= 500:
        segment = "Medium"
    else:
        segment = "Low"

    updated = customer.copy()
    updated["segment"] = segment
    segmented_customers.append(updated)

for customer in segmented_customers:
    print(customer)
```
