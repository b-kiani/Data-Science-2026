### Import pandas

The standard alias is `pd`. Printing the version makes the notebook easier to reproduce.


```python
import pandas as pd

print("pandas version:", pd.__version__)
```

## 1. Series: one labelled sequence

A `Series` contains values plus an index. The index labels help us refer to observations by meaning rather than position.


### Create a Series

The values are daily sales and the index contains day labels.


```python
daily_sales = pd.Series(
    [120, 145, 138],
    index=["Mon", "Tue", "Wed"],
    name="sales_gbp",
)

print(daily_sales)
print("Tuesday:", daily_sales.loc["Tue"])
```

## 2. DataFrame: a labelled table

A `DataFrame` has rows and named columns. A dictionary is a natural way to create one: each key becomes a column name and each list becomes that column's values. All columns must have the same length.


### Create a DataFrame from a dictionary

Each row describes one student. Column names are short but meaningful.


```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104],
    "name": ["Amina", "Ben", "Chen", "Daria"],
    "score": [78, 64, 91, 73],
    "attendance": [0.92, 0.81, 0.97, 0.88],
})

print(students)
```

## 3. Inspect before analysing

Never assume a newly received table is correct. Check its size, labels, types, sample rows, and summary statistics before answering questions.


### Use core inspection tools

`.head()` and `.tail()` return rows; `.shape`, `.columns`, and `.dtypes` describe structure. `.info()` prints a concise audit.


```python
print("First rows:", students.head(2))
print("Last rows:", students.tail(2))
print("Shape:", students.shape)
print("Columns:", students.columns.tolist())
print("Data types:", students.dtypes)
print("Info:")
students.info()
```

### Summarise numeric columns

`.describe()` reports count, mean, spread, and selected percentiles. These summaries are evidence, not a substitute for understanding the data source.


```python
numeric_summary = students.describe()
print(numeric_summary)
```

## 4. Select columns

- `frame["column"]` returns one Series.
- `frame[["column_a", "column_b"]]` returns a DataFrame.

The double brackets in the second form are deliberate: the outer pair selects, and the inner list names multiple columns.


### Choose one or several columns

Print the type so the structural difference is visible.


```python
names = students["name"]
names_and_scores = students[["name", "score"]]

print(names)
print("One-column type:", type(names))
print(names_and_scores)
print("Two-column type:", type(names_and_scores))
```

## 5. Select rows with `.loc` and `.iloc`

- `.loc` uses labels and Boolean conditions.
- `.iloc` uses integer positions.

Both can select rows and columns as `[rows, columns]`.


        ### Compare label and position selection

        The default row labels happen to be integers, but `.loc` and `.iloc` still express different intentions.

<div class="warning"><strong>Take care:</strong> A `.loc` label slice includes its final label, while an `.iloc` position slice excludes its stop position.</div>


```python
print("Row label 1:", students.loc[1])
print("Second row by position:", students.iloc[1])
print("First two rows, name through score:", students.loc[0:1, "name":"score"])
print("First two rows, first three columns:", students.iloc[0:2, 0:3])
```

## 6. Filter rows

The comparison creates a Boolean Series aligned with the DataFrame index. Use it inside `.loc` and request only the columns needed for the result.


### Filter using one condition

This keeps students with a score of at least 75.


```python
high_score_mask = students["score"] >= 75
high_scores = students.loc[high_score_mask, ["name", "score"]]

print(high_score_mask)
print(high_scores)
```

### Filter using two conditions

As with NumPy, parenthesise each comparison and combine them with `&` or `|`.


```python
support_mask = (students["score"] < 75) | (students["attendance"] < 0.85)
support_list = students.loc[support_mask, ["student_id", "name", "score", "attendance"]]

print(support_list)
```

## 7. Create transparent transformations

Use `.copy()` before changing a table you may need later. Then create clearly named columns from existing evidence.


### Add calculated and categorical columns

The source remains unchanged. The new percentage column is rounded for display, and the status is derived from a stated threshold.


```python
student_report = students.copy()
student_report["attendance_pct"] = (student_report["attendance"] * 100).round(1)
student_report["status"] = student_report["score"].apply(
    lambda value: "Pass" if value >= 70 else "Review"
)

print(student_report)
print("Original columns:", students.columns.tolist())
```

### Rename, drop, and sort

These methods return a new DataFrame unless assignment or `inplace=True` is used. Explicit assignment makes each transformation visible.


```python
presentation = student_report.rename(columns={"name": "student_name"})
presentation = presentation.drop(columns=["attendance"])
presentation = presentation.sort_values("score", ascending=False)

print(presentation)
```

## Guided lab: product performance

We will turn a dictionary into a report, inspect it, filter it, calculate revenue, and rank the rows.


### Build a compact product report

The calculation multiplies corresponding columns because pandas aligns them by row index.


```python
products = pd.DataFrame({
    "product": ["Notebook", "Pen", "Folder", "Marker"],
    "price": [3.50, 1.20, 2.25, 1.80],
    "units": [40, 90, 35, 55],
    "category": ["Paper", "Writing", "Paper", "Writing"],
})

product_report = products.copy()
product_report["revenue"] = product_report["price"] * product_report["units"]
strong_products = product_report.loc[
    product_report["revenue"] >= 90,
    ["product", "category", "revenue"],
].sort_values("revenue", ascending=False)

print(product_report)
print("Revenue of at least £90:", strong_products)
```
