## 2. Converting value types

Data from a form or CSV file may arrive as text. Python cannot perform normal numerical arithmetic until suitable text is converted.

- `int()` converts suitable text to a whole number;
- `float()` converts suitable text to a decimal number;
- `str()` converts a value to text.


```python
price_text = "12.50"
quantity_text = "4"

price = float(price_text)
quantity = int(quantity_text)
total = price * quantity

print("Original price type:", type(price_text).__name__)
print("Converted price type:", type(price).__name__)
print("Converted quantity type:", type(quantity).__name__)
print(f"Total: £{total:.2f}")
```

```python
year = 2026
message = "Course year: " + str(year)

print(message)
print(type(message).__name__)
```

<div class="warning"><strong>Conversion can fail:</strong> <code>float("twelve")</code> is invalid because the text does not represent a number. Real data must be checked before conversion.</div>

## 3. Making a decision with `if` and `else`

An `if` statement runs an indented block only when its condition is `True`. An optional `else` block runs when the condition is `False`.

Indentation is part of Python's grammar. The indented lines belong to the decision.


```python
order_total = 32.00

if order_total >= 25:
    print("Free delivery applies")
else:
    print("Delivery fee applies")
```

```python
stock_quantity = 4

if stock_quantity == 0:
    print("Out of stock")
elif stock_quantity < 5:
    print("Low stock: reorder soon")
else:
    print("Stock level is sufficient")
```

`elif` means “otherwise, if this condition is true.” Python checks branches from top to bottom and runs the first matching branch.


```python
score = 67

if score >= 70:
    grade = "Distinction"
elif score >= 60:
    grade = "Pass"
else:
    grade = "Further study required"

print(f"Score: {score}")
print(f"Result: {grade}")
```

Conditions can be combined:

- `and`: both conditions must be true;
- `or`: at least one condition must be true;
- `not`: reverses a Boolean value.


```python
age = 20
registration_complete = True

if age >= 18 and registration_complete:
    print("Student can access the assessment")
else:
    print("Access requirements are not complete")
```

## 4. Your first list

A **list** stores several values in order. Square brackets create the list, and commas separate its items. Week 3 explores lists in depth; today we use only a few essential operations.


```python
weekly_sales = [120, 145, 138, 160, 172]

print(weekly_sales)
print(type(weekly_sales).__name__)
print("Number of days:", len(weekly_sales))
```

Python counts list positions from zero. Therefore the first item is at position `0`.


```python
weekly_sales = [120, 145, 138, 160, 172]

print("First value:", weekly_sales[0])
print("Second value:", weekly_sales[1])
print("Last value:", weekly_sales[-1])
```

```python
total_sales = sum(weekly_sales)
average_sales = total_sales / len(weekly_sales)

print("Total:", total_sales)
print("Average:", average_sales)
print("Minimum:", min(weekly_sales))
print("Maximum:", max(weekly_sales))
print("First-to-last change:", weekly_sales[-1] - weekly_sales[0])
```

```python
# append() adds one value to the end of a list.
weekly_sales.append(181)

print("Updated list:", weekly_sales)
print("Updated number of days:", len(weekly_sales))
print("Updated total:", sum(weekly_sales))
```

## 5. What is data science?

Data science is the disciplined use of data, domain knowledge, code, statistics, and communication to support understanding and decisions.

A typical project asks:

1. What decision or question matters?
2. What data represents the situation?
3. Is that data suitable and trustworthy?
4. What does the analysis show?
5. What are the limitations and risks?
6. What action, if any, is justified?

Complex machine learning is only one possible tool. Clear definitions and reliable data often matter more.


## 6. Structured, semi-structured, and unstructured data

| Data form | Description | Example |
|---|---|---|
| Structured | Consistent rows and columns | Spreadsheet of orders |
| Semi-structured | Labels or keys, but flexible organisation | JSON website event |

## 7. Four levels of question

| Level | Main question | Example |
|---|---|---|
| Descriptive | What happened? | What were total weekly sales? |
| Diagnostic | Where or why? | Which branch had lower sales? |
| Predictive | What may happen next? | Is stock likely to run out? |
| Prescriptive | What should be considered? | How much stock should be ordered? |


```python
# Descriptive: calculate what happened.
monday_sales = 520
tuesday_sales = 610
total_two_days = monday_sales + tuesday_sales

print("Two-day total:", total_two_days)
```

```python
# Diagnostic: compare two locations.
branch_a_sales = 4200
branch_b_sales = 3650
sales_gap = branch_a_sales - branch_b_sales

print("Branch A sales:", branch_a_sales)
print("Branch B sales:", branch_b_sales)
print("Gap (A minus B):", sales_gap)

if branch_a_sales > branch_b_sales:
    print("Branch A recorded higher sales")
else:
    print("Branch B recorded equal or higher sales")
```

```python
# Predictive illustration: a simple rule, not a trained model.
current_stock = 18
average_daily_sales = 6
delivery_wait_days = 4
expected_demand = average_daily_sales * delivery_wait_days
predicted_shortage = expected_demand > current_stock

print("Expected demand before delivery:", expected_demand)
print("Predicted shortage:", predicted_shortage)
```

```python
# Prescriptive illustration: turn the forecast into a possible action.
safety_stock = 5
recommended_order = expected_demand + safety_stock - current_stock

if recommended_order > 0:
    print("Consider ordering", recommended_order, "units")
else:
    print("No additional order is indicated by this rule")
```

<div class="warning"><strong>Careful interpretation:</strong> a calculated recommendation depends on assumptions. The demand estimate may be wrong, delivery may be delayed, and real decisions may involve storage cost, expiry, or supplier limits.</div>

## 8. Data quality: simple first checks

Before calculating, ask whether values are complete, correctly typed, plausible, consistently defined, and free from unintended duplicates.


```python
sale_1 = 120.0
sale_2 = None       # Missing value
sale_3 = "95.50"    # Number stored as text
sale_4 = -40.0      # May be an error or a refund

print("sale_1:", sale_1, type(sale_1).__name__)
print("sale_2:", sale_2, type(sale_2).__name__)
print("sale_3:", sale_3, type(sale_3).__name__)
print("sale_4:", sale_4, type(sale_4).__name__)
```

```python
# Resolve meaning before cleaning. Here we assume:
# - missing sale_2 remains missing and is excluded;
# - sale_3 is verified and converted;
# - sale_4 is a valid refund and remains negative.
verified_sale_3 = float(sale_3)
valid_known_sales = [sale_1, verified_sale_3, sale_4]

print("Known-value total:", sum(valid_known_sales))
print("Known observations:", len(valid_known_sales))
print("Missing observations:", 1)
```

## 9. Guided first analysis

The code below answers a descriptive question about six weeks of revenue, compares two periods, and checks a simple target.


```python
weekly_revenue = [12000, 12800, 12500, 13400, 14100, 14600]
target_per_week = 13000

total_revenue = sum(weekly_revenue)
average_revenue = total_revenue / len(weekly_revenue)
first_three_average = sum(weekly_revenue[0:3]) / 3
last_three_average = sum(weekly_revenue[3:6]) / 3
average_above_target = average_revenue >= target_per_week

print(f"Total revenue: £{total_revenue:,.2f}")
print(f"Average weekly revenue: £{average_revenue:,.2f}")
print(f"First three-week average: £{first_three_average:,.2f}")
print(f"Last three-week average: £{last_three_average:,.2f}")
print(f"Average is at or above target: {average_above_target}")
```

```python
period_change = last_three_average - first_three_average
period_change_percentage = period_change / first_three_average

print(f"Change in three-week average: £{period_change:,.2f}")
print(f"Percentage change: {period_change_percentage:.1%}")

if period_change > 0:
    direction = "increased"
elif period_change < 0:
    direction = "decreased"
else:
    direction = "did not change"

print(f"Finding: average weekly revenue {direction} in the later period.")
```

A responsible conclusion also states a limitation:

> Revenue was higher in the final three weeks than in the first three weeks, but six observations cannot establish a long-term trend or explain the cause.

## 10. What does a data scientist do?

A data scientist does more than write code. Typical responsibilities include:

- understanding the real decision or question;
- defining observations and variables;
- obtaining, documenting, and checking data;
- analysing patterns and uncertainty;
- building and evaluating models when appropriate;
- communicating findings, limitations, and risks;
- monitoring whether a deployed process remains useful and fair.

Python is valuable because it makes an analysis repeatable: the same instructions can be checked and run again on updated data.
