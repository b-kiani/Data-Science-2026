## 1. The Python toolchain

| Component | Plain-language purpose |
|---|---|
| Python | Reads and executes Python instructions |
| Anaconda or Miniconda | Helps install Python and manage environments |
| conda environment | An isolated collection of Python and packages for one project |
| Package | Reusable code written by others, such as NumPy or pandas |
| JupyterLab | Interface containing notebooks, files, terminals, and running kernels |
| Kernel | The active Python process that runs notebook code |

One computer can contain several Python environments. A frequent source of confusion is installing a package in one environment while the notebook uses another.


## 2. Safe installation

Download Anaconda or Miniconda only from the official provider. Use the current 64-bit installer suitable for your operating system.

### Windows

1. Run the official graphical installer for the current user.
2. Keep the default location unless your institution specifies otherwise.
3. Open **Anaconda Prompt** from the Start menu.

### Linux

1. Download the official shell installer.
2. Verify its checksum when required by your institution.
3. Run it in a terminal, initialise conda, and open a new terminal.

<div class="warning"><strong>Security:</strong> do not copy installation commands from unknown websites, and never paste passwords, tokens, or confidential data into a notebook.</div>

### 4. Check the Python version

`sys` is a module from Python's standard library. `sys.version` describes the Python currently running this notebook.


````python
import sys

print("Full Python description:")
print(sys.version)
print("Short version:", sys.version.split()[0])
````

### 5. Check the active executable

The executable path identifies the exact Python program used by the notebook kernel. Keep this information when reporting environment problems.


````python
import sys

python_executable = sys.executable
print("This notebook is using:")
print(python_executable)
````

### 6. Check the operating system

The `platform` module provides information about the computer and operating system.


````python
import platform

print("Operating system:", platform.system())
print("Release:", platform.release())
print("Machine type:", platform.machine())
print("Detailed platform:", platform.platform())
````

### 7. Identify the working folder

The working folder is the default location Python uses for relative file paths. `Path.cwd()` reports it.


````python
from pathlib import Path

working_folder = Path.cwd()
print("Working folder:", working_folder)
print("Folder exists:", working_folder.exists())
````

        ### 8. Construct paths safely

        `pathlib` joins folder and file names using the correct separator for the operating system.

<div class="warning"><strong>Take care:</strong> Constructing paths does not create the folders or file; it only creates path descriptions.</div>


````python
from pathlib import Path

project_folder = Path("course_work")
data_folder = project_folder / "data"
sales_file = data_folder / "sales.csv"

print("Project folder:", project_folder)
print("Data folder:", data_folder)
print("Sales file:", sales_file)
print("File suffix:", sales_file.suffix)
print("File name:", sales_file.name)
````

### 9. Absolute and relative paths

A relative path begins from the working folder. An absolute path identifies a location from the filesystem root.


````python
from pathlib import Path

relative_path = Path("data") / "customers.csv"
absolute_description = relative_path.resolve()

print("Relative:", relative_path)
print("Absolute description:", absolute_description)
print("Does the file currently exist?", relative_path.exists())
````

### 10. Check whether packages are installed

`find_spec` checks whether Python can locate a package in the current environment without importing it.


````python
from importlib.util import find_spec

print("NumPy installed:", find_spec("numpy") is not None)
print("pandas installed:", find_spec("pandas") is not None)
print("SQLAlchemy installed:", find_spec("sqlalchemy") is not None)
print("JupyterLab installed:", find_spec("jupyterlab") is not None)
````

### 11. Display installed package versions

A version number matters because software behaviour can change between releases.


````python
from importlib.util import find_spec

if find_spec("numpy") is not None:
    import numpy
    print("NumPy version:", numpy.__version__)
else:
    print("NumPy is not available in this kernel")

if find_spec("pandas") is not None:
    import pandas
    print("pandas version:", pandas.__version__)
else:
    print("pandas is not available in this kernel")
````

### 12. A simple environment smoke test

A smoke test performs a small calculation to confirm that Python and an installed numerical package work together.


````python
values = [10, 20, 30, 40]
python_average = sum(values) / len(values)
print("Python average:", python_average)

from importlib.util import find_spec
if find_spec("numpy") is not None:
    import numpy as np
    array_average = np.array(values).mean()
    print("NumPy average:", array_average)
else:
    print("NumPy part skipped until the package is installed")
````

## 13. Common setup problems

| Symptom | Likely check |
|---|---|
| `conda` is not recognised | Open Anaconda Prompt or initialise conda correctly |
| Package is installed but import fails | Confirm notebook executable and terminal environment match |
| Notebook shows `[*]` for a long time | Code is still running; interrupt the kernel if appropriate |
| Wrong Python version | Activate the intended environment before launching Jupyter |
| File cannot be found | Check working folder, path spelling, and file existence |

A useful support request includes the operating system, Python version, executable path, environment name, exact command, and complete error message.
