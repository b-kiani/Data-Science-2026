### 1. Create and inspect a list

A list stores several items in order. Square brackets create it and commas separate items.


```python
regions = ["North", "South", "East", "West"]

print(regions)
print(type(regions).__name__)
print("Number of items:", len(regions))
```

### 2. Index positions

Python begins counting at zero. Negative indexes count from the end.


```python
regions = ["North", "South", "East", "West"]

print("First:", regions[0])
print("Second:", regions[1])
print("Last:", regions[-1])
print("Second from last:", regions[-2])
```

### 3. Slice a list

`list[start:stop]` includes the start position but excludes the stop position.


```python
monthly_sales = [82, 91, 88, 104, 110, 107, 116, 121]

print("First three:", monthly_sales[0:3])
print("Also first three:", monthly_sales[:3])
print("Latest three:", monthly_sales[-3:])
print("Positions 2 to 4:", monthly_sales[2:5])
print("Every second value:", monthly_sales[::2])
```

### 4. Replace an item

Lists are mutable: an item can be replaced at a known position.


```python
stock = [12, 8, 5, 10]
print("Before:", stock)

stock[2] = 7
print("After:", stock)
```

### 5. Add items

`append` adds one item at the end; `insert` adds at a chosen position; `extend` adds each item from another collection.


```python
products = ["Basic", "Premium"]
products.append("Enterprise")
print("After append:", products)

products.insert(1, "Standard")
print("After insert:", products)

products.extend(["Student", "Charity"])
print("After extend:", products)
```

        ### 6. Remove items

        `remove` deletes the first matching value. `pop` removes and returns an item, using the final item by default.

<div class="warning"><strong>Take care:</strong> `remove` raises an error if the value is absent. Check membership first when absence is possible.</div>


```python
products = ["Basic", "Standard", "Premium", "Student"]
products.remove("Student")
print("After remove:", products)

removed_product = products.pop()
print("Popped:", removed_product)
print("Remaining:", products)
```

### 7. Membership

`in` and `not in` produce Boolean results.


```python
approved_regions = ["North", "South", "East", "West"]
submitted_region = "South"

print("Approved:", submitted_region in approved_regions)
print("Missing label approved:", "Central" in approved_regions)

if submitted_region in approved_regions:
    print("Region accepted")
else:
    print("Check the region label")
```

### 8. Numerical list summaries

Several built-in functions work directly with numerical lists.


```python
order_values = [120.0, 85.5, 210.0, 99.0, 64.5]

print("Count:", len(order_values))
print("Total:", sum(order_values))
print("Minimum:", min(order_values))
print("Maximum:", max(order_values))
print("Average:", sum(order_values) / len(order_values))
```

### 9. Sort without losing the original

`sorted` returns a new sorted list. `.sort()` changes the original list in place.


```python
scores = [72, 91, 68, 84]
sorted_scores = sorted(scores)

print("Original:", scores)
print("New sorted list:", sorted_scores)

scores.sort(reverse=True)
print("Original after .sort():", scores)
```

### 10. Count and locate values

`.count` reports occurrences; `.index` reports the first matching position.


```python
ratings = [5, 4, 5, 3, 5, 4]

print("Number of 5 ratings:", ratings.count(5))
print("First position of rating 3:", ratings.index(3))
```

### 11. Copy versus alias

Two variable names can refer to the same list. Use `.copy()` when an independent list is needed.


```python
original = [10, 20, 30]
alias = original
independent = original.copy()

alias[0] = 999
independent[1] = 555

print("Original:", original)
print("Alias:", alias)
print("Independent copy:", independent)
```

### 12. Nested lists

A list can contain other lists. Use one index for the inner list and a second for its item.


```python
quarterly_sales = [
    [120, 135, 142],
    [98, 110, 105],
    [160, 155, 172],
]

print("First row:", quarterly_sales[0])
print("First row, second value:", quarterly_sales[0][1])
print("Third row, last value:", quarterly_sales[2][-1])
```

### 13. Tuples

A tuple is ordered but immutable: its items cannot be replaced after creation.


```python
london_coordinates = (51.5074, -0.1278)

print(london_coordinates)
print(type(london_coordinates).__name__)
print("Latitude:", london_coordinates[0])
print("Longitude:", london_coordinates[1])
```

### 14. Tuple unpacking

Unpacking assigns tuple items to separate variable names in one line.


```python
product_record = ("P100", "Notebook", 8.75)
product_id, product_name, product_price = product_record

print("ID:", product_id)
print("Name:", product_name)
print("Price:", product_price)
```

## 15. List or tuple?

| Need | Choose |
|---|---|
| Items will be added, removed, or corrected | List |
| Ordered values should remain fixed | Tuple |
| A changing basket of products | List |
| Fixed latitude/longitude pair | Tuple |

Immutability helps communicate intent, but it does not make a value automatically correct.


### 16. Guided inventory analysis

Combine list creation, indexing, summary functions, appending, and sorting.


```python
weekly_units = [42, 38, 51, 47, 55, 60, 49]

total_units = sum(weekly_units)
average_units = total_units / len(weekly_units)
first_to_last_change = weekly_units[-1] - weekly_units[0]
ordered_units = sorted(weekly_units, reverse=True)

print("Daily units:", weekly_units)
print("Total:", total_units)
print(f"Average: {average_units:.1f}")
print("First-to-last change:", first_to_last_change)
print("Highest to lowest:", ordered_units)
```
