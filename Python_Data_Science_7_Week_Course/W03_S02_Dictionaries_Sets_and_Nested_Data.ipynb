### 1. Create a dictionary

A dictionary stores labelled key–value pairs inside curly brackets. Keys are unique.


```python
customer = {
    "customer_id": "C104",
    "name": "Amina Khan",
    "region": "North",
    "orders": 7,
    "active": True,
}

print(customer)
print(type(customer).__name__)
print("Number of fields:", len(customer))
```

### 2. Retrieve values by key

Unlike a list index, a dictionary key describes the value being retrieved.


```python
print("Customer ID:", customer["customer_id"])
print("Name:", customer["name"])
print("Orders:", customer["orders"])
```

### 3. Use `.get()` safely

Square brackets raise `KeyError` when a key is absent. `.get()` can return a chosen default.


```python
print("Email:", customer.get("email"))
print("Phone:", customer.get("phone", "Not supplied"))
print("Region:", customer.get("region", "Unknown"))
```

### 4. Add and change entries

Assigning a new key adds it. Assigning an existing key replaces its value.


```python
customer["email"] = "amina@example.com"
customer["orders"] = 8

print(customer)
```

### 5. Remove entries

`.pop(key)` removes a key and returns its value. `del` removes a known key without returning it.


```python
record = {"id": 1, "name": "Sample", "temporary_note": "remove me"}
removed_note = record.pop("temporary_note")

print("Removed:", removed_note)
print("Remaining record:", record)
```

### 6. Inspect dictionary views

`.keys()`, `.values()`, and `.items()` provide views of a dictionary's contents.


```python
print("Keys:", list(customer.keys()))
print("Values:", list(customer.values()))
print("Items:", list(customer.items()))
print("Has region key:", "region" in customer)
```

### 7. Iterate through fields

This short `for` loop visits each key–value pair. Loops are taught fully in Week 4.


```python
for key, value in customer.items():
    readable_key = key.replace("_", " ").title()
    print(f"{readable_key}: {value}")
```

### 8. A list of dictionary records

Tabular data is often represented as a list in which every dictionary is one observation.


```python
orders = [
    {"order_id": 501, "region": "North", "total": 120.0},
    {"order_id": 502, "region": "South", "total": 85.5},
    {"order_id": 503, "region": "North", "total": 210.0},
]

print("First record:", orders[0])
print("Second order region:", orders[1]["region"])
print("Third order total:", orders[2]["total"])
```

### 9. Nested dictionaries

Read one level at a time: first the customer field, then a field inside that dictionary.


```python
order = {
    "order_id": 701,
    "customer": {"name": "Leo", "segment": "Premium"},
    "delivery": {"city": "London", "priority": True},
    "total": 145.0,
}

print("Customer name:", order["customer"]["name"])
print("Customer segment:", order["customer"]["segment"])
print("Delivery city:", order["delivery"]["city"])
```

### 10. Nested lists inside dictionaries

A dictionary value may itself be a list.


```python
order = {
    "order_id": 702,
    "items": ["Notebook", "Pen", "Folder"],
    "quantities": [2, 4, 1],
}

print("All items:", order["items"])
print("First item:", order["items"][0])
print("Quantity for second item:", order["quantities"][1])
order["items"].append("Ruler")
print("Updated items:", order["items"])
```

### 11. Create a set

A set stores unique values. It is unordered, so do not rely on display order.


```python
raw_regions = ["North", "South", "North", "West", "South"]
unique_regions = set(raw_regions)

print("Original list:", raw_regions)
print("Unique set:", unique_regions)
print("Unique count:", len(unique_regions))
```

### 12. Add, remove, and check set values

Sets use `.add()` for one value and `.discard()` for safe removal.


```python
approved_regions = {"North", "South", "East"}
approved_regions.add("West")
approved_regions.discard("Central")

print("Approved regions:", approved_regions)
print("North approved:", "North" in approved_regions)
```

### 13. Compare sets

Intersection finds shared values; union finds all values; difference finds values unique to one side.


```python
newsletter = {"C1", "C2", "C3", "C4"}
event_attendees = {"C3", "C4", "C5"}

print("Both groups:", newsletter & event_attendees)
print("Either group:", newsletter | event_attendees)
print("Newsletter only:", newsletter - event_attendees)
print("Event only:", event_attendees - newsletter)
print("Exactly one group:", newsletter ^ event_attendees)
```

### 14. Guided customer summary

Use a list of records and a set to inspect data. The loop is a preview for Week 4.


```python
customers = [
    {"id": "C1", "region": "North", "spend": 420.0, "active": True},
    {"id": "C2", "region": "South", "spend": 180.0, "active": False},
    {"id": "C3", "region": "North", "spend": 610.0, "active": True},
    {"id": "C4", "region": "West", "spend": 250.0, "active": True},
]

customer_ids = {record["id"] for record in customers}
regions = {record["region"] for record in customers}
total_spend = sum(record["spend"] for record in customers)

print("Customer IDs:", customer_ids)
print("Regions:", regions)
print("Total spend:", total_spend)
print("Average spend:", total_spend / len(customers))
```
