### Recall: create and inspect

Begin by recreating a small array. A notebook should show important setup rather than depend on another notebook's memory.


```python
import numpy as np

readings = np.array([4, 7, 9, 12, 15])
print(readings)
print("Shape:", readings.shape)
```

## 1. Indexing one-dimensional arrays

An **index** is a position. Python counts from zero, so the first position is `0`. Negative indices count back from the end.


        ### Select one value

        Square brackets contain the requested position. Predict each result before running it.

<div class="warning"><strong>Take care:</strong> Index `1` is the second item, not the first.</div>


```python
values = np.array([10, 20, 30, 40, 50])

print("First:", values[0])
print("Third:", values[2])
print("Last:", values[-1])
```

### Select a slice

A slice `start:stop` includes the start and excludes the stop. Leaving an end blank means continue to that end of the array.


```python
values = np.array([10, 20, 30, 40, 50])

print("Positions 1 to 3:", values[1:4])
print("First three:", values[:3])
print("From position 2:", values[2:])
print("Every second value:", values[::2])
```

## 2. Indexing two-dimensional arrays

For a table-like array, use `[row, column]`. Both numbers start at zero. A colon means “all positions” on an axis.


### Select cells, rows, and columns

This table has three rows and four columns. Study how each expression relates to the printed output.


```python
scores = np.array([
    [72, 81, 65, 90],
    [88, 79, 91, 84],
    [60, 74, 70, 77],
])

print("Row 1, column 2:", scores[1, 2])
print("First row:", scores[0, :])
print("Third column:", scores[:, 2])
print("Top-left 2 by 2 block:", scores[:2, :2])
```

## 3. Boolean masks

A comparison against an array produces `True` or `False` for every element. This Boolean array is a **mask**. Placing the mask inside square brackets keeps only the `True` positions.


### Create and apply a mask

Separating the condition from the selection makes the reasoning easy to inspect.


```python
response_times = np.array([120, 85, 210, 95, 160])
slow_mask = response_times > 100

print("Mask:", slow_mask)
print("Slow responses:", response_times[slow_mask])
```

        ### Combine conditions

        Use `&` for element-by-element AND and `|` for element-by-element OR. Put each comparison in parentheses.

<div class="warning"><strong>Take care:</strong> Use `&` and `|` with NumPy conditions, not Python's `and` and `or`.</div>


```python
ages = np.array([16, 21, 35, 68, 72, 44])
working_age_mask = (ages >= 18) & (ages < 65)

print("Working-age values:", ages[working_age_mask])
print("Count:", working_age_mask.sum())
```

### Replace values conditionally

`np.where(condition, value_if_true, value_if_false)` creates a new result without changing the source.


```python
raw_scores = np.array([58, 74, 49, 81, 66])
labels = np.where(raw_scores >= 60, "Pass", "Review")

print(labels)
print("Original still numeric:", raw_scores)
```

## 4. Aggregate across rows or columns

An **aggregation** reduces several values to a summary such as a total or mean.

- `axis=0` works down the rows and returns one result per column.
- `axis=1` works across the columns and returns one result per row.


### Compare overall, column, and row means

Name the result according to what it represents, not according to the syntax used to create it.


```python
department_sales = np.array([
    [100, 120, 140],
    [80, 95, 110],
])

overall_mean = department_sales.mean()
mean_per_month = department_sales.mean(axis=0)
mean_per_department = department_sales.mean(axis=1)

print("Overall:", overall_mean)
print("Per month:", mean_per_month)
print("Per department:", mean_per_department)
```

## 5. Broadcasting

Broadcasting lets NumPy combine arrays with compatible shapes. A single value can be applied everywhere; a one-dimensional array can sometimes be matched across the final axis of a table.


### Broadcast a scalar

Adding one value to an array adjusts every element.


```python
prices = np.array([10.0, 15.0, 22.0])
prices_with_fee = prices + 2.5

print(prices_with_fee)
```

        ### Broadcast column adjustments across rows

        The three adjustments align with the three columns of each row.

<div class="warning"><strong>Take care:</strong> Shapes must be compatible. Print both `.shape` values when a broadcasting error occurs.</div>


```python
baseline = np.array([
    [10, 20, 30],
    [40, 50, 60],
])
column_adjustments = np.array([1, 2, 3])

adjusted = baseline + column_adjustments
print(adjusted)
```

## 6. Combine arrays

Concatenation joins existing arrays. The non-joining dimensions must agree.


### Concatenate, stack vertically, and stack horizontally

Choose the operation according to whether the new values represent more observations or more features.


```python
first_week = np.array([12, 15, 14])
second_week = np.array([16, 13, 18])
full_period = np.concatenate([first_week, second_week])

north = np.array([[10, 11], [12, 13]])
south = np.array([[20, 21], [22, 23]])

print("One sequence:", full_period)
print("More rows:", np.vstack([north, south]))
print("More columns:", np.hstack([north, south]))
```

## 7. Missing numeric values

NumPy commonly represents an unavailable decimal value with `np.nan` (“not a number”). Ordinary summary functions can propagate the missing value. `np.nanmean` and related functions explicitly ignore it.


### Detect and summarise missing values

First count missing values, then decide whether ignoring them is appropriate for the question.


```python
sensor_values = np.array([14.2, np.nan, 15.8, 16.1])
missing_mask = np.isnan(sensor_values)

print("Missing mask:", missing_mask)
print("Missing count:", missing_mask.sum())
print("Mean ignoring missing:", np.nanmean(sensor_values))
```

## Guided lab: analyse store traffic

Rows are stores and columns are morning, afternoon, and evening. We will select, filter, adjust, and summarise the same dataset.


### Complete the traffic analysis

The analysis preserves the source array and gives every intermediate result a meaningful name.


```python
visitors = np.array([
    [25, 42, 31],
    [18, 36, 27],
    [33, 48, 40],
])

afternoon = visitors[:, 1]
busy_periods = visitors[visitors >= 40]
totals_per_store = visitors.sum(axis=1)
estimated_next_week = np.rint(visitors * 1.05).astype(int)

print("Afternoon:", afternoon)
print("Periods with at least 40 visitors:", busy_periods)
print("Totals per store:", totals_per_store)
print("5% growth estimate:", estimated_next_week)
```
