## 3. Values and basic data types

Python needs to know what kind of value it is handling.

| Python type | Meaning | Example |
|---|---|---|
| `str` | Text (a string) | `"London"` |
| `int` | Whole number | `25` |
| `float` | Number with a decimal point | `19.95` |
| `bool` | `True` or `False` | `True` |

Text must be quoted. Numbers used for calculations are normally not quoted.


```python
print("London")
print(25)
print(19.95)
print(True)
```

```python
# type() asks Python to report the type of a value.
print(type("London"))
print(type(25))
print(type(19.95))
print(type(True))
```

```python
# Quotation marks change a number into text.
print(type(25))
print(type("25"))

print(25 + 5)       # Numerical addition
print("25" + "5")   # Text is joined together
```

<div class="warning"><strong>Notice:</strong> <code>25</code> and <code>"25"</code> look similar to a person, but Python treats them differently. One is a number; the other is text.</div>

## 4. Variables

A **variable** is a name that refers to a value. The equals sign assigns the value on the right to the name on the left.


```python
student_name = "Amina"
student_age = 20
attendance_rate = 0.90
assignment_complete = True

print(student_name)
print(student_age)
print(attendance_rate)
print(assignment_complete)
```

```python
# Variables can be used more than once.
city = "London"

print(city)
print("The course is taught in", city)
print("Students are travelling to", city)
```

```python
# Assigning a new value changes what the variable refers to.
seats_available = 12
print("Before booking:", seats_available)

seats_available = 11
print("After booking:", seats_available)
```

### Variable naming rules

Good names describe the value: `monthly_revenue` is clearer than `x`.

- Names may contain letters, numbers, and underscores.
- Names cannot begin with a number.
- Spaces are not allowed; use an underscore.
- Python distinguishes uppercase and lowercase.
- Do not use quotation marks around a variable name when referring to it.

Examples: `customer_name`, `order_total`, `week_1_score`.

## 5. Arithmetic operators

| Operator | Operation | Example |
|---|---|---|
| `+` | Addition | `8 + 2` |
| `-` | Subtraction | `8 - 2` |
| `*` | Multiplication | `8 * 2` |
| `/` | Division | `8 / 2` |
| `//` | Whole-number division | `9 // 2` |
| `%` | Remainder | `9 % 2` |
| `**` | Power | `3 ** 2` |

Use parentheses to make the intended order clear.


```python
print("Addition:", 8 + 2)
print("Subtraction:", 8 - 2)
print("Multiplication:", 8 * 2)
print("Division:", 8 / 2)
print("Whole-number division:", 9 // 2)
print("Remainder:", 9 % 2)
print("Power:", 3 ** 2)
```

```python
# Parentheses make the business calculation unambiguous.
unit_price = 12.50
quantity = 4
delivery_fee = 5.00

order_total = (unit_price * quantity) + delivery_fee
print("Order total:", order_total)
```

```python
# Calculate a percentage discount.
original_price = 80.00
discount_rate = 0.15

discount_amount = original_price * discount_rate
final_price = original_price - discount_amount

print("Original price:", original_price)
print("Discount amount:", discount_amount)
print("Final price:", final_price)
```

```python
# Calculate a mean (average).
score_1 = 68
score_2 = 74
score_3 = 82

total_score = score_1 + score_2 + score_3
average_score = total_score / 3

print("Total:", total_score)
print("Average:", average_score)
```

## 6. Working with text

Python calls text a **string**. Strings may use single or double quotation marks, provided the opening and closing marks match.


```python
first_name = "Maya"
last_name = "Patel"

full_name = first_name + " " + last_name
print(full_name)
print("Number of characters:", len(full_name))
```

An **f-string** puts variable values inside readable text. Begin the string with `f` and place each variable or expression inside `{}`.


```python
product = "Notebook"
price = 6.50
quantity = 3
total = price * quantity

print(f"Product: {product}")
print(f"Quantity: {quantity}")
print(f"Total: £{total:.2f}")
```

```python
# Useful string operations return new strings.
raw_department = "  CUSTOMER service  "

cleaned_department = raw_department.strip().title()

print("Raw:", repr(raw_department))
print("Cleaned:", cleaned_department)
print("Lowercase:", cleaned_department.lower())
print("Uppercase:", cleaned_department.upper())
print("Replacement:", cleaned_department.replace("Service", "Support"))
```

## 7. Comparisons and Boolean results

A comparison asks a question and produces `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

A single `=` assigns a value. A double `==` compares two values.


```python
stock_quantity = 7
reorder_level = 10

print(stock_quantity < reorder_level)
print(stock_quantity == 7)
print(stock_quantity != 0)
print(stock_quantity >= 5)
```

```python
course_score = 72
pass_mark = 60
passed = course_score >= pass_mark

print(f"Score: {course_score}")
print(f"Pass mark: {pass_mark}")
print(f"Passed: {passed}")
```
