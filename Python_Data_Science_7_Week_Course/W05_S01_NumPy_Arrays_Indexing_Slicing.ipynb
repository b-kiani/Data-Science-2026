## Recall: calculate with a Python list

A list can store readings. To change every reading, we usually write a loop or a comprehension. Read this code, predict the result, and then run it.


### A list calculation

Each item is visited separately. This is valid Python and makes the repeated work visible.


```python
temperatures_c = [18.0, 20.5, 22.0, 19.5]
temperatures_f = [(value * 9 / 5) + 32 for value in temperatures_c]
print(temperatures_f)
```

## 1. Import NumPy

NumPy is an external package. The conventional alias is `np`, so examples across textbooks and workplaces use the same short name.

If the import fails on your own computer, run `conda install numpy` in Anaconda Prompt, restart the kernel, and try again.


### Import and confirm the version

The version is useful when asking for technical help because behaviour can differ across releases.


```python
import numpy as np

print("NumPy version:", np.__version__)
```

### Create a one-dimensional array

`np.array()` converts a sequence into an array. The printed values look list-like, but the object has NumPy capabilities.


```python
daily_steps = np.array([4200, 6100, 8050, 7300])

print(daily_steps)
print(type(daily_steps))
```

        ### Lists and arrays use `+` differently

        List addition joins containers. Array addition performs arithmetic on corresponding elements.

<div class="warning"><strong>Take care:</strong> Do not assume an operator means exactly the same thing for every data type. Check the type and test a small example.</div>


```python
python_values = [1, 2, 3]
numpy_values = np.array([1, 2, 3])

print("List + list:", python_values + python_values)
print("Array + array:", numpy_values + numpy_values)
```

## 2. Inspect an array

Four attributes answer different questions:

| Attribute | Question answered |
|---|---|
| `.ndim` | How many axes are there? |
| `.shape` | How long is each axis? |
| `.size` | How many values are there altogether? |
| `.dtype` | What kind of value is stored? |


### Inspect one-dimensional data

A one-dimensional array has one axis. Its shape is written as a one-item tuple, so `(4,)` means four positions on one axis.


```python
scores = np.array([68, 74, 81, 90])

print("Dimensions:", scores.ndim)
print("Shape:", scores.shape)
print("Number of values:", scores.size)
print("Data type:", scores.dtype)
```

### Inspect a two-dimensional array

A table-like array has rows and columns. Here the shape `(2, 3)` means two rows and three columns.


```python
weekly_sales = np.array([
    [12, 15, 11],
    [18, 17, 20],
])

print(weekly_sales)
print("Dimensions:", weekly_sales.ndim)
print("Shape:", weekly_sales.shape)
print("Total values:", weekly_sales.size)
```

## 3. Create common arrays

NumPy can create useful starting arrays without typing every element. Shapes are supplied as tuples such as `(2, 3)`.


### Zeros, ones, and a chosen value

These constructors are useful for initialising a calculation or creating predictable test data.


```python
zero_grid = np.zeros((2, 3))
one_grid = np.ones((2, 3))
missing_code_grid = np.full((2, 3), -1)

print("Zeros:", zero_grid)
print("Ones:", one_grid)
print("Chosen value:", missing_code_grid)
```

        ### Evenly spaced sequences

        `arange(start, stop, step)` excludes the stop value. `linspace(start, stop, count)` includes both ends and creates a requested number of points.

<div class="warning"><strong>Take care:</strong> The second argument to `arange` is a stopping boundary, not necessarily the final value produced.</div>


```python
hours = np.arange(0, 24, 6)
proportions = np.linspace(0, 1, 5)

print("Hours:", hours)
print("Proportions:", proportions)
```

## 4. Data types and conversion

An array normally uses one data type for all values. NumPy selects a suitable type, or we can request one. Converting from decimal numbers to integers removes the fractional part; it does not perform ordinary rounding.


### Choose or convert a data type

`dtype=float` creates decimal storage. `.astype(int)` returns a new integer array.


```python
counts_as_decimal = np.array([2, 4, 6], dtype=float)
measurements = np.array([3.9, 5.2, 7.8])
whole_measurements = measurements.astype(int)

print(counts_as_decimal, counts_as_decimal.dtype)
print(whole_measurements, whole_measurements.dtype)
```

## 5. Vectorised calculations

When an operation is applied to an array, NumPy usually applies it element by element. This style is called **vectorisation**. It is concise and is normally faster than a Python loop for large numeric datasets.


### Apply one formula to every value

The formula is written once, yet all four temperatures are converted.


```python
temperatures_c_array = np.array([18.0, 20.5, 22.0, 19.5])
temperatures_f_array = (temperatures_c_array * 9 / 5) + 32

print(temperatures_f_array)
```

### Combine corresponding values

Arrays of compatible shapes can be combined position by position.


```python
morning_sales = np.array([10, 12, 9, 15])
afternoon_sales = np.array([8, 11, 13, 10])
daily_totals = morning_sales + afternoon_sales

print("Daily totals:", daily_totals)
print("Grand total:", daily_totals.sum())
print("Mean per day:", daily_totals.mean())
```

## 6. Reshape without changing the values

`reshape()` changes the arrangement, not the number of elements. A six-value array can become `2 × 3` or `3 × 2`, but not `4 × 2`.


### Turn a sequence into a table

The original and reshaped arrays contain the same six values.


```python
sequence = np.arange(1, 7)
table = sequence.reshape(2, 3)

print("Original:", sequence)
print("Reshaped:", table)
print("New shape:", table.shape)
```

### Make an independent copy

`.copy()` is useful when the original data must remain unchanged while we test a transformation.


```python
original = np.array([10, 20, 30])
experiment = original.copy()
experiment[0] = 999

print("Original:", original)
print("Experiment:", experiment)
```

## Guided lab: a small air-quality dataset

We have particulate readings for two sites across three days. We will inspect the structure, convert the units, and calculate a short report.


### Build and summarise the dataset

Rows represent sites and columns represent days. Descriptive names keep the meaning visible.


```python
pm25_micrograms = np.array([
    [11.2, 14.5, 12.1],
    [18.0, 16.8, 20.4],
])

pm25_milligrams = pm25_micrograms / 1000

print("Shape:", pm25_micrograms.shape)
print("Overall mean:", round(pm25_micrograms.mean(), 2))
print("Highest reading:", pm25_micrograms.max())
print("Values in mg/m³:", pm25_milligrams)
```
