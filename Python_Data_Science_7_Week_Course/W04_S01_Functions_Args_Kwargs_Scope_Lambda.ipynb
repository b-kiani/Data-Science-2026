### 1. Define and call a function

`def` creates a named block of reusable code. The indented body runs only when the function is called.


```python
def greet_student():
    print("Welcome to the Python class")

greet_student()
greet_student()
```

### 2. Add a parameter

A parameter is a name in the function definition. An argument is the value supplied during a call.


```python
def greet_student(name):
    print(f"Welcome, {name}")

greet_student("Amina")
greet_student("Leo")
```

### 3. Return a result

`return` sends a value back to the caller. A returned value can be stored and reused.


```python
def calculate_revenue(price, quantity):
    revenue = price * quantity
    return revenue

first_revenue = calculate_revenue(12.50, 4)
second_revenue = calculate_revenue(8.00, 7)

print("First:", first_revenue)
print("Second:", second_revenue)
print("Combined:", first_revenue + second_revenue)
```

### 4. Printing is not returning

A function that only prints displays information but sends back `None`.


```python
def print_total(amount):
    print(f"Total: £{amount:.2f}")

result = print_total(25.00)
print("Returned value:", result)
```

### 5. Positional and keyword arguments

Positional arguments depend on order. Keyword arguments name the parameter and can improve clarity.


```python
def order_summary(product, quantity, price):
    return f"{quantity} × {product} at £{price:.2f}"

print(order_summary("Notebook", 3, 7.50))
print(order_summary(product="Pen", price=1.25, quantity=10))
```

### 6. Default parameter values

A default is used when the caller omits that argument.


```python
def final_price(price, discount_rate=0.0):
    return price * (1 - discount_rate)

print("No discount:", final_price(100))
print("Ten percent discount:", final_price(100, 0.10))
print("Keyword argument:", final_price(price=80, discount_rate=0.25))
```

### 7. Validate function inputs

A function can reject invalid values with a clear exception.


```python
def average_revenue(total_revenue, customers):
    if customers <= 0:
        raise ValueError("customers must be greater than zero")
    return total_revenue / customers

print(average_revenue(1200, 8))
# average_revenue(1200, 0) would raise the intended ValueError.
```

### 8. Return more than one value

Python packs multiple returned values into a tuple, which can be unpacked.


```python
def range_summary(values):
    smallest = min(values)
    largest = max(values)
    return smallest, largest

minimum_value, maximum_value = range_summary([12, 8, 19, 14])
print("Minimum:", minimum_value)
print("Maximum:", maximum_value)
```

### 9. Local scope

A variable created inside a function is normally local to that call.


```python
tax_rate = 0.20

def add_tax(amount, rate):
    taxed_amount = amount * (1 + rate)
    return taxed_amount

print(add_tax(50, tax_rate))
# taxed_amount is local and is not available here outside the function.
```

### 10. `*args`: flexible positional arguments

`*args` collects additional positional arguments into a tuple.


```python
def total_values(*values):
    print("Collected type:", type(values).__name__)
    print("Collected values:", values)
    return sum(values)

print("Two values:", total_values(10, 20))
print("Four values:", total_values(10, 20, 30, 40))
```

### 11. Keyword-only options after `*args`

A named option after `*args` must be passed by keyword, improving readability.


```python
def discounted_total(*prices, discount=0.0):
    subtotal = sum(prices)
    return subtotal * (1 - discount)

print(discounted_total(10, 20, 5))
print(discounted_total(10, 20, 5, discount=0.10))
```

### 12. `**kwargs`: flexible keyword arguments

`**kwargs` collects additional named arguments into a dictionary.


```python
def create_profile(name, **details):
    profile = {"name": name}
    profile.update(details)
    return profile

student = create_profile("Maya", programme="Business", year=1, active=True)
print(student)
print("Details type:", type(student).__name__)
```

### 13. Combine required, default, `*args`, and `**kwargs`

The parameter order below is common: required values, flexible positional values, keyword-only defaults, then flexible keyword metadata.


```python
def build_order(customer_id, *prices, discount=0.0, **metadata):
    subtotal = sum(prices)
    final_total = subtotal * (1 - discount)
    return {
        "customer_id": customer_id,
        "subtotal": subtotal,
        "final_total": final_total,
        "metadata": metadata,
    }

order = build_order("C10", 12.5, 8.0, 4.5,
                    discount=0.10, channel="online", priority=True)
print(order)
```

### 14. Lambda functions

A lambda is a small anonymous function containing one expression. Use it for a short local operation, not complex logic.


```python
prices = [45, 210, 18, 75]
sorted_prices = sorted(prices, key=lambda price: price)
print(sorted_prices)

products = [
    {"name": "Keyboard", "price": 45},
    {"name": "Monitor", "price": 210},
    {"name": "Mouse", "price": 18},
]
by_price = sorted(products, key=lambda product: product["price"])
print(by_price)
```

### 15. Modules and packages

Modules contain reusable Python code. Packages organise multiple modules. Imports provide access without copying implementations into the notebook.


```python
import math
import statistics as stats
from datetime import date

print("Square root:", math.sqrt(81))
print("Mean:", stats.mean([5, 8, 11]))
print("Date:", date.today())
```

### 16. Guided reusable sales calculation

This function validates inputs, uses a default tax rate, and returns a dictionary of reusable results.


```python
def calculate_invoice(unit_price, quantity, discount_rate=0.0, tax_rate=0.20):
    if unit_price < 0 or quantity < 0:
        raise ValueError("price and quantity cannot be negative")
    subtotal = unit_price * quantity
    discount = subtotal * discount_rate
    after_discount = subtotal - discount
    tax = after_discount * tax_rate
    return {
        "subtotal": subtotal,
        "discount": discount,
        "tax": tax,
        "total": after_discount + tax,
    }

invoice = calculate_invoice(25.0, 4, discount_rate=0.10)
print(invoice)
print(f"Amount due: £{invoice['total']:.2f}")
```
