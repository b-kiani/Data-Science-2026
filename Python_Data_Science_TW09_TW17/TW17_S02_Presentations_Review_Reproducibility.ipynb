## Presentation structure: seven evidence-led slides

1. decision, audience, and why it matters;
2. dataset, unit of analysis, scope, and provenance;
3. quality issues and cleaning decisions;
4. exploratory finding with one clear chart;
5. method, baseline, and validation design;
6. result, errors, limitations, and ethics;
7. recommendation and next test.

Aim for one message per slide. Put detailed tables and code in an appendix or notebook.


### Create a timed storyboard

The total presentation is seven minutes, leaving time for questions.


```python
import numpy as np
import pandas as pd

storyboard = pd.DataFrame({
    "slide": range(1, 8),
    "purpose": [
        "Decision and question",
        "Data and scope",
        "Quality and cleaning",
        "Exploratory evidence",
        "Method and validation",
        "Results and limitations",
        "Recommendation and next test",
    ],
    "seconds": [45, 45, 60, 75, 75, 75, 45],
})
print(storyboard)
print("Total minutes:", storyboard["seconds"].sum() / 60)
```

## 1. Present evidence, not a notebook tour

Do not narrate every code cell. State the decision, show only evidence needed to support it, explain one important limitation, and end with a proportionate next action.


### Convert technical output into an executive evidence table

The language separates estimate, uncertainty, and operational meaning.


```python
evidence = pd.DataFrame([
    {
        "claim": "Candidate improves on majority baseline F1",
        "evidence": "Held-out F1 comparison",
        "business_meaning": "The model identifies some churners rather than predicting all retained",
        "limitation": "Small synthetic dataset",
    },
    {
        "claim": "Lower threshold increases recall",
        "evidence": "Threshold sensitivity table",
        "business_meaning": "More possible churners found, but more unnecessary contacts",
        "limitation": "Error costs require stakeholder input",
    },
])
print(evidence)
```

## 2. Reproducibility audit

Another person should be able to extract the folder, install requirements, restart the kernel, run all cells in order, and obtain the stated tables and metrics without hidden manual steps.


### Audit notebook evidence

This generic checklist can be completed during peer review.


```python
reproducibility_audit = pd.DataFrame({
    "check": [
        "Data source path is relative",
        "Random seeds are fixed",
        "Raw data is preserved",
        "Cleaning order is documented",
        "Train/test separation occurs before learned preprocessing",
        "Metrics are calculated rather than typed",
        "Figures have titles, labels, and units",
        "Notebook runs from a clean kernel",
        "Requirements are documented",
    ],
    "status": ["Pass", "Pass", "Pass", "Review", "Pass", "Pass", "Pass", "Review", "Pass"],
    "evidence_or_action": ["", "", "", "Add decision log", "", "", "", "Peer must rerun", ""],
})
print(reproducibility_audit)
```

## 3. Statistical and model audit

Ask whether summaries fit variable types, missingness choices are justified, uncertainty is visible, correlations avoid causal wording, model metrics match the decision, and the final test was not repeatedly used for tuning.


### Create a claim-strength ladder

Match wording to design strength.


```python
claim_ladder = pd.DataFrame({
    "design": [
        "Descriptive observational analysis",
        "Predictive model with external validation",
        "Randomised controlled experiment",
    ],
    "supported_wording": [
        "is associated with / was higher in this dataset",
        "predicts with measured performance in the validated setting",
        "caused an average change under the studied intervention",
    ],
    "unsupported_shortcut": [
        "causes",
        "will work everywhere",
        "proves no harms or subgroup differences",
    ],
})
print(claim_ladder)
```

## 4. Ethical and operational audit

Review legitimate purpose, data minimisation, privacy, affected groups, subgroup errors, human oversight, appeal routes, deployment monitoring, drift, and what happens when the model is wrong.


### Score risk controls

A low score identifies missing work; it is not a certification.


```python
ethics_controls = pd.Series({
    "legitimate_purpose_defined": 1,
    "data_minimisation_reviewed": 1,
    "sensitive_and_proxy_features_reviewed": 0,
    "subgroup_performance_checked": 0,
    "human_oversight_defined": 1,
    "monitoring_and_stop_rule_defined": 0,
})
print(ethics_controls)
print("Controls evidenced:", f"{ethics_controls.mean():.0%}")
```

## 5. Peer-review rubric

Feedback should be specific, evidenced, and actionable. Replace “good chart” with “the title states the finding, but the y-axis needs pounds and the sample size should be shown.”


### Build a 100-point rubric

Weights prioritise reasoning and reproducibility rather than visual polish alone.


```python
rubric = pd.DataFrame({
    "criterion": [
        "Question and decision",
        "Data provenance and quality",
        "Statistical and EDA reasoning",
        "Preprocessing and leakage control",
        "Model or segmentation validation",
        "Visual communication",
        "Limitations, ethics, and recommendation",
        "Reproducibility and teamwork",
    ],
    "maximum_points": [10, 15, 15, 15, 15, 10, 12, 8],
})
print(rubric)
print("Total points:", rubric["maximum_points"].sum())
```

## 6. Module review: the end-to-end chain

Every step constrains the next:

business decision → data and unit → import and provenance → quality and cleaning → statistics and EDA → features → validation design → model or analysis → limitations → recommendation → monitoring.


### Check key concepts with executable examples

The dictionary acts as a compact revision map.


```python
revision_map = {
    "nominal_example": "region",
    "robust_centre": "median",
    "linear_association": "Pearson correlation",
    "safe_unknown_category": "OneHotEncoder(handle_unknown='ignore')",
    "classification_baseline": "DummyClassifier",
    "cluster_separation_metric": "silhouette score",
    "variance_projection": "PCA",
    "time_safe_validation": "chronological holdout / TimeSeriesSplit",
}
for concept, example in revision_map.items():
    print(f"{concept}: {example}")
```

### Run a small integrity quiz

Assertions reveal whether key statements are internally consistent.


```python
review_answers = {
    "probabilities_are_0_to_1": True,
    "correlation_proves_causation": False,
    "fit_scaler_before_split": False,
    "test_set_is_for_repeated_tuning": False,
    "cluster_labels_are_arbitrary": True,
    "random_split_is_default_for_time_series": False,
}

assert review_answers["probabilities_are_0_to_1"]
assert not review_answers["correlation_proves_causation"]
assert not review_answers["fit_scaler_before_split"]
assert not review_answers["test_set_is_for_repeated_tuning"]
assert review_answers["cluster_labels_are_arbitrary"]
assert not review_answers["random_split_is_default_for_time_series"]
print("Integrity quiz passed.")
```
