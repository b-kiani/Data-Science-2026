### Import scientific tools

SciPy provides tested distribution and hypothesis-test functions. The aliases keep formulas readable.


```python
import numpy as np
import pandas as pd
from scipy import stats

print("SciPy loaded:", stats is not None)
```

## 1. Binomial distribution

Use a binomial model when there is a fixed number of independent trials, each trial has two outcomes, and the success probability is constant.

Example: among 10 independently contacted customers, how many respond when each has a 30% response probability?


### Calculate exact and cumulative binomial probabilities

The PMF gives probability at an exact discrete value. The CDF gives probability at or below a value.


```python
n_trials = 10
p_success = 0.30

p_exactly_four = stats.binom.pmf(4, n_trials, p_success)
p_at_most_four = stats.binom.cdf(4, n_trials, p_success)
p_at_least_four = 1 - stats.binom.cdf(3, n_trials, p_success)

print("P(X = 4):", round(p_exactly_four, 4))
print("P(X <= 4):", round(p_at_most_four, 4))
print("P(X >= 4):", round(p_at_least_four, 4))
```

## 2. Poisson distribution

Use a Poisson model for counts of events in a fixed interval when events occur independently at an approximately constant average rate. The rate parameter is commonly written as lambda.


        ### Model calls arriving in one hour

        If the historical mean is 3.2 calls per hour, the PMF can estimate an exact count and the CDF a range.

<div class="warning"><strong>Take care:</strong> The model is inappropriate when the event rate changes sharply, events influence one another, or counts have much more variability than a Poisson model permits.</div>


```python
hourly_rate = 3.2
p_exactly_five = stats.poisson.pmf(5, mu=hourly_rate)
p_more_than_five = 1 - stats.poisson.cdf(5, mu=hourly_rate)

print("P(exactly 5 calls):", round(p_exactly_five, 4))
print("P(more than 5 calls):", round(p_more_than_five, 4))
```

## 3. Normal distribution, PDF, and CDF

The normal distribution is continuous and symmetric. Its PDF describes density, not the probability at one exact point. For a continuous variable, probability is an area over an interval, often found by subtracting CDF values.


### Find probability over a normal interval

Assume delivery time is approximately normal with mean 30 minutes and standard deviation 5 minutes.


```python
mean_minutes = 30
sd_minutes = 5

density_at_30 = stats.norm.pdf(30, loc=mean_minutes, scale=sd_minutes)
p_under_35 = stats.norm.cdf(35, loc=mean_minutes, scale=sd_minutes)
p_between_25_and_35 = (
    stats.norm.cdf(35, loc=mean_minutes, scale=sd_minutes)
    - stats.norm.cdf(25, loc=mean_minutes, scale=sd_minutes)
)

print("PDF at 30:", round(density_at_30, 4))
print("P(X < 35):", round(p_under_35, 4))
print("P(25 < X < 35):", round(p_between_25_and_35, 4))
```

### Standardise a value with a z-score

A z-score states how many standard deviations a value is above or below the mean.


```python
observed_minutes = 42
z_score = (observed_minutes - mean_minutes) / sd_minutes
upper_tail_probability = 1 - stats.norm.cdf(observed_minutes, mean_minutes, sd_minutes)

print("z-score:", z_score)
print("P(X >= 42):", round(upper_tail_probability, 4))
```

## 4. Pearson correlation

Pearson correlation r measures the strength and direction of a **linear** relationship between two numerical variables.

- r is between -1 and 1;
- the sign gives direction;
- magnitude gives linear strength;
- r does not prove causation;
- outliers and non-linear patterns can mislead it.


### Calculate correlation and inspect the paired values

Correlation needs paired observations. Always inspect a scatter plot in a real analysis, even when a coefficient is available.


```python
advertising = np.array([5, 8, 10, 12, 15, 18, 22, 25])
sales = np.array([20, 24, 25, 31, 35, 38, 45, 48])

r_value, correlation_p = stats.pearsonr(advertising, sales)
print("Pearson r:", round(r_value, 3))
print("Correlation p-value:", round(correlation_p, 6))
```

### See how an outlier can alter correlation

One unusual point can substantially change a small dataset.


```python
sales_with_outlier = sales.copy()
sales_with_outlier[-1] = 5

original_r = stats.pearsonr(advertising, sales).statistic
outlier_r = stats.pearsonr(advertising, sales_with_outlier).statistic

print("Original r:", round(original_r, 3))
print("With outlier r:", round(outlier_r, 3))
```

## 5. Hypothesis testing

A test begins before looking at the p-value:

1. define the population question;
2. state a null hypothesis H0 and alternative H1;
3. choose a test and significance level;
4. check assumptions;
5. calculate the statistic and p-value;
6. report effect size, uncertainty, and practical importance.

A p-value is the probability of data at least this incompatible with H0 **assuming H0 and the test model are true**. It is not the probability that H0 is true.


### Run an independent two-sample t-test

The Welch version does not assume equal population variances. We also report the mean difference as an interpretable effect.


```python
group_a = np.array([72, 75, 78, 74, 77, 79, 73, 76])
group_b = np.array([68, 71, 70, 72, 69, 74, 67, 71])

test_result = stats.ttest_ind(group_a, group_b, equal_var=False)
mean_difference = group_a.mean() - group_b.mean()

print("Mean A:", round(group_a.mean(), 2))
print("Mean B:", round(group_b.mean(), 2))
print("Mean difference:", round(mean_difference, 2))
print("t statistic:", round(test_result.statistic, 3))
print("p-value:", round(test_result.pvalue, 4))
```

### Construct a confidence interval for one mean

A 95% interval gives a range produced by a method that captures the true mean in 95% of repeated samples under its assumptions.


```python
sample_values = np.array([42, 45, 39, 47, 44, 46, 41, 43, 48, 40])
sample_mean = sample_values.mean()
standard_error = stats.sem(sample_values)
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(sample_values) - 1,
    loc=sample_mean,
    scale=standard_error,
)

print("Mean:", round(sample_mean, 2))
print("95% CI:", tuple(round(value, 2) for value in confidence_interval))
```

## Guided analysis: an A/B service experiment

The question is whether average completion time differs between two independently assigned processes. We report sample sizes, descriptive statistics, mean difference, and a Welch t-test. A causal claim would additionally require credible random assignment and a well-run experiment.


### Report statistical and practical evidence

A small p-value alone is not enough. The observed difference must matter in the operational context.


```python
process_a = np.array([31, 29, 34, 30, 32, 28, 33, 31, 29, 30])
process_b = np.array([27, 25, 29, 26, 28, 24, 27, 25, 26, 28])
alpha = 0.05

result = stats.ttest_ind(process_a, process_b, equal_var=False)
difference = process_a.mean() - process_b.mean()
decision = "reject H0" if result.pvalue < alpha else "do not reject H0"

print("n A, n B:", len(process_a), len(process_b))
print("Means:", round(process_a.mean(), 2), round(process_b.mean(), 2))
print("A minus B:", round(difference, 2), "minutes")
print("p-value:", round(result.pvalue, 5))
print("Decision at alpha=0.05:", decision)
```
