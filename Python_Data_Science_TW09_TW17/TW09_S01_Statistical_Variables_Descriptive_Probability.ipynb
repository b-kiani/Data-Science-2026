### Recall: one row, one variable, one value

A variable is a recorded characteristic. Its type determines which calculations have a sensible interpretation.


```python
import numpy as np
import pandas as pd

sample = pd.DataFrame({
    "customer_id": ["C01", "C02", "C03", "C04"],
    "region": ["North", "South", "North", "West"],
    "satisfaction": ["Low", "High", "Medium", "High"],
    "orders": [1, 4, 2, 5],
    "spend": [24.5, 88.0, 41.5, 102.0],
})
print(sample)
```

## 1. Types of statistical variables

| Type | Meaning | Example | Useful summaries |
|---|---|---|---|
| Nominal | categories with no natural order | region | counts, proportions, mode |
| Ordinal | ordered categories with unequal or unknown gaps | low, medium, high | counts, median category |
| Discrete | countable numerical values | number of orders | mean, median, frequency |
| Continuous | measured values on a scale | spend, height, time | mean, median, standard deviation |

An identifier may contain digits but is still a nominal label. Calculating the average customer ID would have no business meaning.


### Create a small variable catalogue

A data dictionary records the intended meaning and type instead of guessing from appearance.


```python
variable_catalogue = pd.DataFrame({
    "variable": ["customer_id", "region", "satisfaction", "orders", "spend"],
    "statistical_type": ["Nominal", "Nominal", "Ordinal", "Discrete", "Continuous"],
})
print(variable_catalogue)
```

## 2. Measures of centre

- The **mean** uses every value and is sensitive to extreme observations.
- The **median** is the middle ordered value and is more resistant to extremes.
- The **mode** is the most frequent value and can describe categories.

No measure is automatically best. Choose the summary that answers the question and fits the data.


### Compare mean, median, and mode

The income outlier pulls the mean upward, while the median remains near the typical values.


```python
incomes = pd.Series([28, 30, 31, 32, 34, 35, 120], name="income_thousands")

print("Mean:", incomes.mean())
print("Median:", incomes.median())
print("Mode values:", incomes.mode().tolist())
```

### See an outlier change the mean

The same first six values are summarised with and without the extreme observation.


```python
without_outlier = incomes[incomes < 100]
comparison = pd.DataFrame({
    "measure": ["mean", "median"],
    "with_outlier": [incomes.mean(), incomes.median()],
    "without_outlier": [without_outlier.mean(), without_outlier.median()],
})
print(comparison)
```

## 3. Spread and position

Centre alone can hide important differences. Spread tells us how dispersed the observations are.

- range = maximum minus minimum;
- variance = average squared distance from the mean;
- standard deviation = square root of variance, in the original unit;
- quartiles divide ordered data into four sections;
- interquartile range (IQR) = Q3 minus Q1.

pandas uses the sample standard deviation by default, with one degree-of-freedom correction.


### Calculate a compact descriptive report

Use named values so the result can be interpreted without remembering array positions.


```python
delivery_minutes = pd.Series([22, 25, 25, 28, 30, 34, 36, 41])
q1 = delivery_minutes.quantile(0.25)
q3 = delivery_minutes.quantile(0.75)

summary = {
    "count": delivery_minutes.count(),
    "mean": delivery_minutes.mean(),
    "median": delivery_minutes.median(),
    "range": delivery_minutes.max() - delivery_minutes.min(),
    "variance": delivery_minutes.var(),
    "standard_deviation": delivery_minutes.std(),
    "q1": q1,
    "q3": q3,
    "iqr": q3 - q1,
}
print(pd.Series(summary).round(2))
```

        ### Use the IQR rule as a flag, not an automatic deletion rule

        The rule marks unusually distant observations. Domain knowledge decides whether they are errors or genuine cases.

<div class="warning"><strong>Take care:</strong> A statistically unusual value is not automatically wrong. Investigate its source before changing it.</div>


```python
values = pd.Series([10, 11, 12, 13, 13, 14, 15, 40])
q1 = values.quantile(0.25)
q3 = values.quantile(0.75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
flagged = values[(values < lower_fence) | (values > upper_fence)]

print("Fences:", lower_fence, upper_fence)
print("Flagged values:", flagged.tolist())
```

## 4. Probability language

A probability lies from 0 to 1. An **event** is an outcome or collection of outcomes.

- complement: event A does not happen;
- intersection: A and B both happen;
- union: A or B or both happen;
- conditional probability: probability of A given that B is known.

For equally likely outcomes, probability is favourable outcomes divided by possible outcomes.


### Calculate simple event probabilities

A six-sided die has six equally likely outcomes. The even event contains three outcomes.


```python
outcomes = {1, 2, 3, 4, 5, 6}
even = {2, 4, 6}
greater_than_three = {4, 5, 6}

p_even = len(even) / len(outcomes)
p_not_even = 1 - p_even
p_even_and_gt3 = len(even & greater_than_three) / len(outcomes)
p_even_or_gt3 = len(even | greater_than_three) / len(outcomes)

print("P(even):", p_even)
print("P(not even):", p_not_even)
print("P(even and >3):", p_even_and_gt3)
print("P(even or >3):", p_even_or_gt3)
```

### Calculate conditional probability from a table

The denominator changes: among Premium customers only, how many renewed?


```python
retention = pd.DataFrame({
    "plan": ["Basic", "Basic", "Premium", "Premium", "Premium", "Premium"],
    "renewed": [0, 1, 1, 1, 0, 1],
})
premium = retention[retention["plan"] == "Premium"]
p_renewed_given_premium = premium["renewed"].mean()

print("Premium observations:", len(premium))
print("P(renewed | Premium):", p_renewed_given_premium)
```

### Estimate probability by simulation

A fixed random seed makes the result reproducible. Larger simulations usually move closer to the theoretical probability.


```python
rng = np.random.default_rng(42)
simulated_dice = rng.integers(1, 7, size=10_000)
estimated_probability = np.mean(simulated_dice >= 5)

print("Estimated P(result >= 5):", round(estimated_probability, 4))
print("Theoretical probability:", round(2 / 6, 4))
```

## Guided analysis: compare two service teams

We will classify variables, compare centre and spread, and estimate the probability that a randomly selected response time is under 30 minutes.


### Create an evidence-based comparison

The result reports sample size as well as summaries, because a mean without its context can mislead.


```python
service = pd.DataFrame({
    "team": ["A"] * 8 + ["B"] * 8,
    "minutes": [21, 24, 25, 26, 28, 31, 35, 50, 20, 22, 25, 27, 29, 30, 32, 34],
})

team_summary = service.groupby("team").agg(
    observations=("minutes", "count"),
    mean_minutes=("minutes", "mean"),
    median_minutes=("minutes", "median"),
    sd_minutes=("minutes", "std"),
)
p_under_30 = (service["minutes"] < 30).mean()

print(team_summary.round(2))
print("P(random response is under 30 minutes):", round(p_under_30, 3))
```
